```
# Check that the 3.4 s rest span before each cue stays clear of the previous trial's imagery (cue + 4 s).
import sys, numpy as np
sys.path.insert(0, 'src')
from bci_glove.data import load_subject_continuous, subject_list
rows = []
for s in subject_list():
    raw, events, _ = load_subject_continuous(s, session='1test')
    gaps = np.diff(events[:, 0]) / raw.info['sfreq']
    gaps = gaps[gaps < 60]  # ignore breaks between runs
    rows.append((s, gaps.min()))
    print(f'subject {s}: shortest cue-to-cue gap {gaps.min():.2f} s')
print('Rest span is clean' if min(g for _, g in rows) >= 7.4 else 'WARNING: rest span overlaps the previous trial')
```
